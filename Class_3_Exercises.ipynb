## Loops
Loops in Python are used to execute a block of code repeatedly. We will look at two different types of loops; the *for loop*, and the *while loop*.  
**The for loop** iterates over sequences, so it keeps executing until it has reached the end of the loop.  
**The while loop** keeps repeating while a condition is true, and first ends when the condition becomes false. 

```python
# For loop - Basic syntax
# for element in sequence:
    # code to execute for each element

# Example of for loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(len(fruit))
```

```python
# While loop - Basic syntax
# while condition:
    # code to execute while condition is True
    
# Example of while loop
x = 1
while x <= 5:
    print(x)
    x += 1
```

## Password checker
You’re making a simple password checker. How would you check if the password is correct?

Write code that asks for a password and checks if it follows the requirements:
- It needs to be between 8 and 32 characters long 

Make sure you print some meaningful message the user can understand. Try it with both the correct password and wrong passwords.

Hint: use the input function to allow for user inputs e.g. **input("Enter the password: ")**

```python
# code here 
def Password_func(Password):
    if len(Password) > 32:
        print("Your password must not exceed 32 characters, try another")
    elif len(Password) < 8:
        print("Your password must not be at least 8 characters")
    else:
        print("Password accepted")
```

Now lets add some extra requirements:
- The password needs to have at least 3 digits
- And there needs to be an uppercase letter

```python
# Modified code here
def Password_func_2(Password):
    Requirements_counter = 0 #We put this in for displaying it at the end, there are probably cleaner ways to do it, but this works quite well.
    #The first part here is for checking the length of the password, as we did above.
    if len(Password) > 32:
        print("Your password must not exceed 32 characters, try another")
    elif len(Password) < 8:
        print("Your password must not be at least 8 characters")
    else:
        Requirements_counter +=1 #This will come in handy later

    #The next part is for checking the number of digits
    Digit_count = 0
    for character in Password: #This goes through the characters in the password
        if character.isdigit(): #.isdigit() is for checking if a string only contains digits, and so by going through them one by one, we check for each character.
            Digit_count += 1 #This is effectively a counter of how many digits are in the password
    if Digit_count < 3: #This loop makes sure that only if there are three or more digits, the requirements_counter goes up, otherwise the error message.
        print("Your password needs to have at least three digits")
    else:
        Requirements_counter +=1

    #This part is for checking uppercase letters
    Uppercase_count = 0
    for character in Password: #This is the same logic as before, so see digit counter:)
        if character.isupper(): #.isupper() does the same as .isdigit, except for checking for uppercase. The rest is the same as above.
            Uppercase_count += 1
    if Uppercase_count < 1:
        print("Your password needs at least one uppercase letter")
    else: 
        Requirements_counter +=1
        
    #This is for doing a final message
    if Requirements_counter == 3: #This is why we did the requirements_counter, only if all three requirements are met, does it print "Password accepted".
        print("Password accepted")
    else:
        print("Try again:)")
```

```python
Password_func_2("jwjSwa32212")
```

And the last bits, now we also need to ensure that the password contains:
- 2 special characters

Make sure the user understands why a wrong password is rejected. And ensure that only a password that holds all requirements are passed. 

```python
# Last modifications on the code - final password_checker
def Password_func_3(Password):
    Requirements_counter = 0
    #The first part here is for checking the length
    if len(Password) > 32:
        print("Your password must not exceed 32 characters, try another")
    elif len(Password) < 8:
        print("Your password must not be at least 8 characters")
    else:
        Requirements_counter +=1

    #The next part is for checking the number of digits
    Digit_count = 0
    for character in Password:
        if character.isdigit():
            Digit_count += 1
    if Digit_count < 3:
        print("Your password needs to have at least three digits")
    else:
        Requirements_counter +=1

    #This part is for checking uppercase letters
    Uppercase_count = 0
    for character in Password:
        if character.isupper():
            Uppercase_count += 1
    if Uppercase_count < 1:
        print("Your password needs at least one uppercase letter")
    else: 
        Requirements_counter +=1

    #Here, we now add the special characters checker, it works by and large the same as the other checkers.
    Special_character_list = ["*", "@", "!", "$", "#", "%"] #Since there is no special character function like .isdigit() or .isupper()(i think) we need to specify them 
    Special_character_count = 0
    for character in Password:
        if character in Special_character_list: #This is what we do because we made the list outselves and did not use .is__()
            Special_character_count += 1
    if Special_character_count < 2:
        print(f"Your password needs to include at least 2 special characters, you can choose from {Special_character_list}") #We also tell them the list of characters.
    else:
        Requirements_counter += 1
        
    #This is for doing a final message
    if Requirements_counter == 4: #Note it needs to be four and not three now
        print("Password accepted")
    else:
        print("Try again:)")
```

Well done! 
Now wouldn't it be neat if we could just wrap that into a function that we could call temperature_converter with two parameters "temperature" and "unit" so that we could convert both ways. 



```python
# Temperature converter here
def Temperature_func_3(Temp, Unit):
    if Unit == "Celcius": #We can simply use a conditional for checking units and performing different operations depending on the input unit.
        Result = (Temp * 9/5) + 32
        return(Result)
    elif Unit == "Fahrenheit":
        Result = (Temp - 32)/(9/5)
        return(Result)
    else:
        print("You must put either \"Celcius\" or \"Fahrenheit\" as unit") #The \ are for writing "" inside a string without it not working.
```

## Grade calculator

The scenario: You’re helping a teacher who wants to calculate student grades quickly. The teacher has been doing these calculations by hand, which takes a long time and sometimes leads to mistakes.

The challenge: 
- Calculate the average of test scores
- Convert percentage grades to letter grades (it's just easier than the danish grading metric ngl)
- Calculate final grades with different weights for different assignments

Example data:  
Tests count for 70%, homework counts for 30%  

Test scores: 85, 93, 78, 96, 88
Homework average: 90

Now we start with one thing first. Each functions should do _one_ thing effectively. 


```python
# Make a function that calculate the average of test scores. 
def calculate_average(*args): #Remember to put *args because we do not know how many arguments we are getting.
    Result = sum(args)/len(args)
    return(Result)
```

```python
calculate_average(45, 3, 234, 4)
```

Well done, now we move on to the next sub-task

Hint: Grade to lettter grade converter  
90–100 → A  
80–89  → B  
70–79  → C  
60–69  → D  
0–59   → F  


```python
# Make a function that converst the test scores to letter grades

def percentage_to_letter_grade(Test_Score):
    if Test_Score >= 90: #We use a conditional to see which grade they got and print a message.
        print(f"Gongrats, with a score of {Test_Score} you aced the test and got an A")
    elif Test_Score < 90 and Test_Score >= 80:
        print(f"Nice job, your score of {Test_Score} got you a B")
    elif Test_Score < 80 and Test_Score >= 70:
        print(f"Not bad, your grade is C, with a score of {Test_Score}")
    elif Test_Score < 70 and Test_Score >= 60:
        print(f"You got a D, with a score of {Test_Score}, it's okay, but could be better")
    elif Test_Score < 60 and Test_Score > 0:
        print(f"You got an F, better luck next time, you still got this! See if you can beat your score of {Test_Score} next time:)")
    else:
        print("Please provide a test score in the right format")
```

```python
# Test with different percentages
percentage_to_letter_grade(50)
```

Last task, We need to calculate the weighted grade


```python
import numpy as np #We need a function from this package
def calculate_weighted_grade(Homework_average, *args):
    Test_average = sum(args)/len(args)
    Average_list = [Test_average, Homework_average] #This is something we eed in a bit
    Weight = [70, 30] #We need to make this for something just below
    Weighted_grade = np.average(Average_list, weights = Weight) #Here, we use the function from numpy, which includes weights.
    return(f"Your final weighted score is {int(Weighted_grade)}")
```

```python
calculate_weighted_grade(90, 85, 93, 78, 96, 88)
```

Now we want to give the professor one function that uses all these three and wraps them in a neat little function with everything put together. 

```python
import numpy as np
def calculate_student_grade(Homework_average, *args):
    Test_average = sum(args)/len(args)
    Average_list = [Test_average, Homework_average]
    Weight = [70, 30]
    Weighted_grade = np.average(Average_list, weights = Weight) #We use the numpy function again

    if Weighted_grade >= 90: #Depending on the scores, the grade will change.
        Final_Grade = "A"
    elif Weighted_grade < 90 and Weighted_grade >= 80:
        Final_Grade = "B"
    elif Weighted_grade < 80 and Weighted_grade >= 70:
        Final_Grade = "C"
    elif Weighted_grade < 70 and Weighted_grade >= 60:
        Final_Grade = "D"
    elif Weighted_grade < 60 and Weighted_grade > 0:
        Final_Grade = "F"
    else:
        print("The format seems to be a bit off, try again (remember to put the homework average first and then the test scores:)")

    #Below is a fancy message for them
    print(f"Your test scores are {args}, averaging at {Test_average}\n \nYour average homework score is {Homework_average}\n \nYour final weighted score is {Weighted_grade}\n \nYour final grade is {Final_Grade} \n")
    #And now it also ends with a personalised message depending on the final grade.
    if Weighted_grade >= 90:
        print("Excellent job, you did very well")
    elif Weighted_grade < 90 and Weighted_grade >= 80:
        print("Nice job, quite a good score")
    elif Weighted_grade < 80 and Weighted_grade >= 70:
        print("Not bad, but you can try to do even better next time")
    elif Weighted_grade < 70 and Weighted_grade >= 60:
        print("Not the best score, but it's okay")
    elif Weighted_grade < 60 and Weighted_grade > 0:
        print("Try again next time, we believe in you!")
    else:
        print("The format seems to be a bit off, try again (remember to put the homework average first and then the test scores:)")
```

```python
calculate_student_grade(10, 10, 99, 56, 97, 95)
```
